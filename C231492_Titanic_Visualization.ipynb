```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 100
```

```python
df = pd.read_csv("titanic_data/train.csv")

df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8, 6))

sorted_age = np.sort(df['Age'].dropna())
quantiles = np.arange(1, len(sorted_age) + 1) / len(sorted_age)

plt.plot(quantiles, sorted_age, marker='.', linestyle='none', color='steelblue')

plt.xlabel('Quantiles')
plt.ylabel('Age')
plt.title('Quantile Plot of Age')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

stats.probplot(df['Age'].dropna(), dist="norm", plot=plt)

plt.title('Q-Q Plot of Age')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

stats.probplot(df['Fare'].dropna(), dist="norm", plot=plt)

plt.title('Q-Q Plot of Fare')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', alpha=0.7)

plt.title('Scatter Plot: Age vs Fare (Color: Survived)')
plt.xlabel('Age')
plt.ylabel('Fare (log scale)')
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='Pclass', y='Fare', hue='Survived', alpha=0.7)

plt.title('Scatter Plot: Pclass vs Fare (Color: Survived)')
plt.xlabel('Passenger Class')
plt.ylabel('Fare (log scale)')
plt.yscale('log')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Pclass', y='Survived',
            palette='viridis', hue='Pclass', legend=False)

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
sex_counts = df['Sex'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(sex_counts.values,
        labels=sex_counts.index,
        autopct='%1.1f%%',
        colors=['#3498db', '#e91e63'],
        startangle=90)

plt.title('Gender Distribution of Passengers')
plt.show()
```

```python
plt.figure(figsize=(10, 8))

numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f',
            linewidths=0.5)

plt.title('Correlation Heatmap of Titanic Dataset')
plt.show()
```

```python
target = 'Survived'

corr_matrix = df.select_dtypes(include=['int64','float64']).corr()

positive_features = corr_matrix[target][
    (corr_matrix[target] > 0) &
    (corr_matrix[target].index != target)
].sort_values(ascending=False)

print("Features Positively Correlated with Survived:")
print(positive_features)
```

```python
for feature in positive_features.index:

    plt.figure(figsize=(7, 5))

    sns.scatterplot(x=df[feature], y=df[target], alpha=0.3)

    plt.title(f'{feature} vs {target}')
    plt.xlabel(feature)
    plt.ylabel('Survived')

    plt.show()
```

```python
for feature in positive_features.index:

    plt.figure(figsize=(7, 5))

    sns.boxplot(x=df[target], y=df[feature],
                palette='Set2', hue=df[target], legend=False)

    plt.title(f'{feature} by Survived')
    plt.xlabel('Survived (0=No, 1=Yes)')
    plt.ylabel(feature)

    plt.show()
```

```python
numeric_features = df.select_dtypes(include=['int64','float64']).columns
numeric_features = [f for f in numeric_features if f != target]

for feature in numeric_features:

    plt.figure(figsize=(7, 5))

    sns.scatterplot(x=df[feature], y=df[target], alpha=0.3)

    plt.title(f'{feature} vs {target}')
    plt.xlabel(feature)
    plt.ylabel('Survived')

    plt.show()
```
