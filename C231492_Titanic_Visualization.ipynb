```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 100
```

```python
plt.figure(figsize=(8, 6))

sorted_age = np.sort(df['Age'].dropna())
quantiles = np.arange(1, len(sorted_age) + 1) / len(sorted_age)

plt.plot(quantiles, sorted_age, marker='.', linestyle='none', color='steelblue')

plt.xlabel('Quantiles')
plt.ylabel('Age')
plt.title('Quantile Plot of Age')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

stats.probplot(df['Age'].dropna(), dist="norm", plot=plt)

plt.title('Q-Q Plot of Age')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))

stats.probplot(df['Fare'].dropna(), dist="norm", plot=plt)

plt.title('Q-Q Plot of Fare')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', alpha=0.7)

plt.title('Scatter Plot: Age vs Fare (রঙ: Survived)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(data=df, x='Pclass', y='Fare', hue='Survived', alpha=0.7)

plt.title('Scatter Plot: Pclass vs Fare (রঙ: Survived)')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 8))

numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f',
            linewidths=0.5)

plt.title('Correlation Heatmap of Titanic Dataset')
plt.show()
```

```python

corr_matrix = df.select_dtypes(include=['int64', 'float64']).corr()

positive_corr = corr_matrix['Survived'][corr_matrix['Survived'] > 0].sort_values(ascending=False)

print("Features Positively Correlated with Survived:")
print(positive_corr)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Pclass', y='Survived', palette='viridis')

plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
sex_counts = df['Sex'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(sex_counts.values,
        labels=sex_counts.index,
        autopct='%1.1f%%',
        colors=['#3498db', '#e91e63'],
        startangle=90)
plt.title('Gender Distribution of Passengers')
plt.show()
```
